```python
import pandas as pd

# Dataset load karo
df = pd.read_csv('../data/raw/telco.csv')

# Basic info dekho
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
# Pehli 5 rows dekho
df.head()
```

```python
# Data types aur missing values dekho
df.info()
```

```python
# Target column check karo
df['Churn Label'].value_counts()
```

```python
# Sab columns aur unke missing values dekho
pd.set_option('display.max_rows', None)
print(df.isnull().sum())
```

```python
# Sab column names ek saath dekho (poori list)
for col in df.columns:
    print(col)
```

```python
# Total columns count
print(f"Total columns: {len(df.columns)}\n")

# Sab columns numbered format mein
for i, col in enumerate(df.columns, 1):
    print(f"{i}. {col}")
```

```python
# Sirf woh columns dikhao jinme missing values hain
missing = df.isnull().sum()
missing = missing[missing > 0]
print(missing)
```

```python
# Verify: Internet Type missing sirf unke liye jinke paas Internet Service nahi hai
print(df[df['Internet Type'].isnull()]['Internet Service'].value_counts())
```

```python
# Verify: Churn Label ke against missing Churn Reason
print(df[df['Churn Reason'].isnull()]['Churn Label'].value_counts())
```

```python
# 1. Internet Type - jaha Internet Service "No" hai, wahan "No Internet Service" fill karo
df['Internet Type'] = df['Internet Type'].fillna('No Internet Service')

# 2. Offer - jinko koi offer nahi mila, unke liye "No Offer" fill karo
df['Offer'] = df['Offer'].fillna('No Offer')

# Verify - ab koi missing values nahi honi chahiye in dono mein
print("Internet Type missing:", df['Internet Type'].isnull().sum())
print("Offer missing:", df['Offer'].isnull().sum())
```

```python
# Duplicate rows check karo
print("Duplicate rows:", df.duplicated().sum())

# Customer ID mein bhi duplicates check karo (har customer unique hona chahiye)
print("Duplicate Customer IDs:", df['Customer ID'].duplicated().sum())
```

```python
# Data types dekho
print(df.dtypes)
```

```python
# Sab "object" (text) columns ke unique values dekho
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    print(f"\n{col}: {df[col].unique()}")
```

```python
# Columns jo drop karne hain (identifiers, geography, leakage)
columns_to_drop = [
    'Customer ID', 'Country', 'State', 'City', 'Zip Code', 
    'Latitude', 'Longitude', 'Quarter', 'Under 30',
    'Customer Status', 'Churn Score', 'Satisfaction Score'
]

df_clean = df.drop(columns=columns_to_drop)

print("Original shape:", df.shape)
print("After dropping:", df_clean.shape)
print("\nRemaining columns:")
print(df_clean.columns.tolist())
```

```python
# churn distribution (overall)

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,4))
sns.countplot(data=df_clean, x='Churn Label', palette='Set2')
plt.title('Churn Distribution')
plt.show()
```

```python
# churn by contract type 

plt.figure(figsize=(8,5))
sns.countplot(data=df_clean, x='Contract', hue='Churn Label', palette='Set2')
plt.title('Churn by Contract Type')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df_clean, x='Tenure in Months', hue='Churn Label', multiple='stack', bins=30)
plt.title('Churn by Tenure')
plt.show()
```

```python
# 4. Monthly Charges vs Churn (Price sensitivity dekhne ke liye)
plt.figure(figsize=(8,5))
sns.boxplot(data=df_clean, x='Churn Label', y='Monthly Charge', palette='Set2')
plt.title('Monthly Charges by Churn Status')
plt.show()
```

```python
# . Internet Type vs Churn (Service quality ka effect)
plt.figure(figsize=(8,5))
sns.countplot(data=df_clean, x='Internet Type', hue='Churn Label', palette='Set2')
plt.title('Churn by Internet Type')
plt.xticks(rotation=15)
plt.show()
```

```python
# Payment Method vs Churn (Payment friction ka effect)

plt.figure(figsize=(8,5))
sns.countplot(data=df_clean, x='Payment Method', hue='Churn Label', palette='Set2')
plt.title('Churn by Payment Method')
plt.xticks(rotation=15)
plt.show()
```

```python
#  Churn Category Breakdown (RAG ke liye SABSE zaroori — dekhte hain kaunse reasons sabse zyada hain)

plt.figure(figsize=(8,5))
df_clean['Churn Category'].value_counts().plot(kind='bar', color='coral')
plt.title('Churn Reasons - Category Wise')
plt.xlabel('Category')
plt.ylabel('Count')
plt.xticks(rotation=20)
plt.show()
```

```python
# Correlation Heatmap (Numeric features ke beech relation dekhne ke liye)

plt.figure(figsize=(12,8))
numeric_df = df_clean.select_dtypes(include=['int64', 'float64'])
sns.heatmap(numeric_df.corr(), annot=False, cmap='coolwarm')
plt.title('Correlation Heatmap (Numeric Features)')
plt.show()
```

```python
# Cleaned dataset save karo
df_clean.to_csv('../data/processed/telco_cleaned.csv', index=False)
print("Saved successfully!")
print("Final shape:", df_clean.shape)
```
